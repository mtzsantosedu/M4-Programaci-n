```python
# Identificar variables categóricas y numéricas
categoricas = variable.select_dtypes(include=["object", "category", "bool"]).columns
numericas = variable.select_dtypes(include="number").columns

# Imputar variables categóricas con la moda
for columna in categoricas:
    if variable[columna].isnull().any():
        moda = variable[columna].mode()
        if not moda.empty:
            variable[columna] = variable[columna].fillna(moda.iloc[0])

# Imputar variables numéricas con la media
for columna in numericas:
    if variable[columna].isnull().any():
        media = variable[columna].mean()
        if pd.notna(media):
            variable[columna] = variable[columna].fillna(media)

# Verificar valores perdidos restantes
print(variable.isnull().sum())
```
