```python
from matplotlib.backends.backend_pdf import PdfPages
import textwrap

import matplotlib.pyplot as plt

archivo_pdf = "reporte_actividad.pdf"

lineas = [
    "Reporte de exploración de transacciones",
    "",
    f"Registros: {actividad.shape[0]}",
    f"Columnas: {actividad.shape[1]}",
    "",
    "Columnas del DataFrame:",
    ", ".join(actividad.columns),
    "",
    "Estadísticas de variables numéricas:",
    actividad.describe().round(2).to_string(),
    "",
    "Valores nulos por columna:",
    actividad.isnull().sum().to_string(),
    "",
    "Frecuencia de Categoria_Comercio:",
    actividad["Categoria_Comercio"].value_counts(dropna=False).to_string(),
    "",
    "Distribución de Canal_Pago:",
    actividad["Canal_Pago"].value_counts(dropna=False).to_string(),
    "",
    "Categoria_Comercio cruzada con Es_Fraude:",
    pd.crosstab(
        actividad["Categoria_Comercio"],
        actividad["Es_Fraude"],
        margins=True
    ).to_string(),
]

# Ajustar el ancho de las líneas para que quepan en el PDF
lineas_ajustadas = []
for linea in lineas:
    lineas_ajustadas.extend(textwrap.wrap(linea, width=95) or [""])

with PdfPages(archivo_pdf) as pdf:
    lineas_por_pagina = 42

    for inicio in range(0, len(lineas_ajustadas), lineas_por_pagina):
        pagina = lineas_ajustadas[inicio:inicio + lineas_por_pagina]

        fig = plt.figure(figsize=(8.27, 11.69))  # Tamaño A4
        fig.text(
            0.07, 0.95,
            "\n".join(pagina),
            ha="left",
            va="top",
            family="monospace",
            fontsize=8
        )
        pdf.savefig(fig, bbox_inches="tight")
        plt.close(fig)

print(f"PDF generado: {archivo_pdf}")
```
