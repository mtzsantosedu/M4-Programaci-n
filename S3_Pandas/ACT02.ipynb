## Importancia de identificar los valores perdidos

En un análisis exploratorio de datos (EDA), identificar los valores perdidos permite conocer qué información falta, en qué columnas aparece y con qué frecuencia. Esto es importante porque los datos ausentes pueden:

- **Reducir la cantidad de información disponible** para los análisis y modelos.
- **Sesgar los resultados** si la ausencia no ocurre al azar; por ejemplo, si ciertos grupos tienen más datos faltantes que otros.
- **Afectar cálculos y visualizaciones**, como promedios, correlaciones y distribuciones.
- **Revelar problemas en la recolección o integración de datos**, como errores de captura o fuentes incompletas.

En Pandas, se pueden revisar con:

```python
df.isnull().sum()
```

Después de identificarlos, conviene decidir cómo tratarlos según su cantidad, ubicación y posible causa. Algunas opciones son eliminar registros o columnas, o completar los valores con una estrategia adecuada. Es importante no sustituirlos automáticamente sin evaluar cómo podría afectar el análisis.

Es importante identificar los datos perdidos en un dataframe. Aunque en algunos casos son omitidos automáticamente, muchos procedimientos requieren que se tenga una matriz completa sin datos perdidos. 
- Pandas utiliza el símbolo *NaN* (Not a number) para indicar datos numéricos perdidos (específicamente, datos perdidos de punto flotante). Si son datos de series de tiempo (*datatime*) son indicados como *NaT*.
- Una forma rápida de consultar los datos perdidos en un dataframe es mediante el método *info()*. Al utilizarlo se mostrará el nombre de cada columna y el total de registros (*non-null values*)

# 9. Evaluación de datos perdidos


## Buenas prácticas al tratar datos faltantes

Antes de decidir si **eliminar** o **imputar** (rellenar) los valores faltantes (`NaN`), es importante comprender **por qué se producen**. En estadística y ciencia de datos, los datos faltantes se clasifican en tres mecanismos principales:

### 1. MCAR — *Missing Completely at Random*

**Datos faltantes completamente aleatorios**

* **Explicación:** La probabilidad de que un dato falte **no tiene relación con el valor de ninguna variable**, ni de la propia columna ni de otras variables del conjunto de datos. La ausencia se produce por un factor completamente aleatorio.
* **Ejemplo:** Un fallo momentáneo del sensor provoca que algunas mediciones no se registren, independientemente de las características de los datos.
* **Impacto:** La eliminación de estas observaciones o su imputación mediante técnicas sencillas, como la **media o la mediana**, generalmente no introduce sesgos sistemáticos relacionados con el mecanismo de ausencia.

---

### 2. MAR — *Missing at Random*

**Datos faltantes aleatorios condicionados a variables observadas**

* **Explicación:** La probabilidad de que un dato falte **depende de otra variable que sí está observada**, pero no directamente del valor que falta.
* **Ejemplo:** En una encuesta, los hombres responden con menor frecuencia la variable `frecuencia_uso`. En este caso, la ausencia del dato está relacionada con `genero`, una variable que sí conocemos.
* **Impacto:** Pueden utilizarse técnicas de **imputación multivariada**, en las que otras variables ayudan a estimar el valor faltante. 

---

### 3. MNAR — *Missing Not at Random*

**Datos faltantes no aleatorios**

* **Explicación:** La probabilidad de que un dato falte **depende directamente del valor que no observamos**.
* **Ejemplo:** Las personas con ingresos muy altos podrían tener una mayor probabilidad de no reportar su `ingreso_mensual_usd`. En este caso, la ausencia está relacionada con el propio nivel de ingresos que falta.
* **Impacto:** Eliminar estas observaciones puede generar un **sesgo importante**, ya que se perdería información relevante de un determinado grupo. En estos casos, puede ser conveniente crear primero una **variable indicadora de valores faltantes** y posteriormente aplicar una estrategia de imputación adecuada.

---

### Resumen

| Mecanismo | ¿De qué depende que falte el dato?                                 | Ejemplo                                            | Estrategia posible                                  |
| --------- | ------------------------------------------------------------------ | -------------------------------------------------- | --------------------------------------------------- |
| **MCAR**  | Del azar; no depende de variables observadas ni del valor faltante | Fallo aleatorio de un sensor                       | Eliminación o imputación simple                     |
| **MAR**   | De otras variables observadas                                      | `frecuencia_uso` falta dependiendo de `genero`     | Imputación multivariada o modelos que manejan nulos |
| **MNAR**  | Del propio valor que falta                                         | Personas con ingresos altos no reportan su ingreso | Indicador de nulo + imputación especializada        |

> **Importante:** En la práctica, determinar si los datos son MCAR, MAR o MNAR no siempre es sencillo. La decisión debe apoyarse en el conocimiento del proceso que generó los datos, el análisis exploratorio y, cuando sea posible, pruebas estadísticas y conocimiento del dominio.

# Imputación simple


## Imputación simple: media, mediana y moda

La **imputación simple** consiste en reemplazar los valores faltantes (`NaN`) de una variable por un valor calculado a partir de los datos disponibles. Es una forma sencilla de completar un conjunto de datos, aunque puede alterar su distribución y no siempre es adecuada.

- **Media:** promedio de los valores. Se usa principalmente con variables numéricas y distribuciones sin valores extremos importantes. Puede verse muy afectada por valores atípicos.
- **Mediana:** valor central al ordenar los datos. Se usa con variables numéricas cuando hay valores atípicos o una distribución asimétrica; suele ser más robusta que la media.
- **Moda:** valor que aparece con mayor frecuencia. Se usa principalmente con variables categóricas, aunque también puede aplicarse a variables numéricas discretas.

En Pandas, por ejemplo:

```python
# Media
df["columna"] = df["columna"].fillna(df["columna"].mean())

# Mediana
df["columna"] = df["columna"].fillna(df["columna"].median())

# Moda
df["columna"] = df["columna"].fillna(df["columna"].mode().iloc[0])
```

La estrategia debe elegirse según el tipo de variable, su distribución y el motivo por el que faltan los datos. Conviene calcular estos valores usando únicamente los datos de entrenamiento cuando se preparan datos para un modelo, para evitar filtraciones de información.

````python
# Imputación simple en df3 en la variables numericas la media.
# Eliminar filas completamente vacías y completar los valores numéricos con la media
df3 = df3.dropna(how="all")

columnas_numericas = df3.select_dtypes(include="number").columns
df3[columnas_numericas] = df3[columnas_numericas].fillna(
    df3[columnas_numericas].mean()
)

df3[columnas_numericas].isna().sum()

````

````python
# Imputación simple en df3 en las variables categóricas con la moda
columnas_categoricas = df3.select_dtypes(include="object").columns
df3[columnas_categoricas] = df3[columnas_categoricas].apply(
    lambda col: col.fillna(col.mode().iloc[0])
)

df3[columnas_categoricas].isna().sum()
````

## Importancia del boxplot en el análisis exploratorio

El **boxplot** (diagrama de caja y bigotes) permite resumir visualmente la distribución de una variable numérica y detectar posibles valores atípicos. En la exploración de datos, ayuda a identificar características que conviene revisar antes de realizar análisis estadísticos o entrenar modelos.

Sus elementos principales son:

- **Mediana:** línea dentro de la caja; representa el valor central.
- **Cuartiles:** los bordes de la caja delimitan el 50 % central de los datos.
- **Rango intercuartílico (RIC):** distancia entre el tercer y el primer cuartil.
- **Bigotes y puntos externos:** muestran la dispersión y los valores que podrían ser atípicos.

Al comparar boxplots de varias variables, se pueden observar diferencias en dispersión, asimetría y presencia de valores extremos. Sin embargo, un valor atípico no necesariamente es un error: debe investigarse considerando el contexto de los datos antes de eliminarlo o modificarlo. Además, las variables binarias, como `target_churn`, suelen aportar poca información mediante este tipo de gráfico.

````python
# librerías para visualización
import matplotlib.pyplot as plt
import math

columnas_numericas_df3 = df3.select_dtypes(include="number").columns
n_columnas = 4
n_filas = math.ceil(len(columnas_numericas_df3) / n_columnas)

fig, axes = plt.subplots(n_filas, n_columnas, figsize=(16, n_filas * 3))
axes = axes.flatten()

for ax, columna in zip(axes, columnas_numericas_df3):
    ax.hist(df3[columna].dropna(), bins=20, edgecolor="black")
    ax.set_title(columna)
    ax.set_xlabel("Valor")
    ax.set_ylabel("Frecuencia")

for ax in axes[len(columnas_numericas_df3):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
````

### Exploración y tratamiento de datos

> Carga el archivo en un DataFrame nuevo, por ejemplo, `df04`.

1. **Carga e inspección inicial**  

    Archivo: `04_imputación_de_datos.xlsx`. Consulta sus dimensiones, las primeras y últimas filas y los tipos de datos.  

    **Responde:** ¿Cuántas filas y columnas tiene? ¿Qué columnas son numéricas y cuáles categóricas?

2. **Identificación de valores faltantes**  

    Calcula el total y el porcentaje de valores nulos por columna. Muestra únicamente las columnas que tienen datos faltantes.  

    **Responde:** ¿Qué columna tiene el mayor porcentaje de valores faltantes? ¿Hay filas completamente vacías?

3. **Imputación de valores faltantes**  

    Elimina las filas completamente vacías; después, completa los nulos de las columnas numéricas y categóricas utilizando la imputación simple.  

    **Responde:** ¿Cuántos valores nulos quedan? ¿Por qué puede ser conveniente usar la mediana en vez de la media en algunos casos?

4. **Histogramas de variables numéricas**  

    Elabora un gráfico con las variables numéricas del dataset.

    **Responde:** Describe el comportamiento de las variables de acuerdo con lo que se presenta en tus gráficos. Identifica los principales hallazgos.

5. **Detección de posibles valores atípicos**  

    Elabora diagramas de caja (*boxplots*) para las variables numéricas. Elige una variable con puntos fuera de los bigotes y revisa sus valores extremos.  

    **Responde:** ¿Qué valores podrían ser atípicos? ¿Por qué no se deben eliminar sin investigar primero su contexto?
