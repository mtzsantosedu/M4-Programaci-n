## Introducción a Pandas

**Pandas** es una librería de Python diseñada para trabajar con datos estructurados. Sus principales estructuras son:

- **`Series`**: una secuencia unidimensional de datos con etiquetas.
- **`DataFrame`**: una tabla bidimensional con filas y columnas, similar a una hoja de cálculo o una tabla de base de datos.

Con Pandas se pueden cargar datos desde archivos como CSV y Excel, explorarlos, limpiarlos, transformarlos y analizarlos.

## Relevancia en la ciencia de datos

Pandas es una herramienta fundamental en la ciencia de datos porque facilita la preparación y el análisis de información. Permite, entre otras tareas:

- Filtrar, ordenar y seleccionar datos.
- Tratar valores faltantes y duplicados.
- Agrupar datos y calcular estadísticas.
- Combinar tablas de distintas fuentes.
- Preparar datos para visualizarlos o utilizarlos en modelos de aprendizaje automático.

Su integración con librerías como **NumPy**, **Matplotlib** y **scikit-learn** hace que forme parte de muchas etapas del trabajo con datos en Python.
