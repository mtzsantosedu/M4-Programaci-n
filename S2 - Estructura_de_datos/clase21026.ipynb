```python
# Información de un producto

producto = str(input("Dame el nombre del producto: "))
precio = float(input("Dame el precio: "))
cantidad = int(input("Dame la cantidad: "))

print("Producto:", producto)
print("Precio:", precio)
print("Cantidad:", cantidad)

# Información del segundo producto

producto2 = str(input("Dame el nombre del segundo producto: "))
precio2 = float(input("Dame el precio del segundo producto: "))
cantidad2 = int(input("Dame la cantidad del segundo producto: "))

print("Segundo producto:", producto2)
print("Precio:", precio2)
print("Cantidad:", cantidad2)

# Información del tercer producto

producto3 = str(input("Dame el nombre del tercer producto: "))
precio3 = float(input("Dame el precio del tercer producto: "))
cantidad3 = int(input("Dame la cantidad del tercer producto: "))

print("Tercer producto:", producto3)
print("Precio:", precio3)
print("Cantidad:", cantidad3)
```

```python
# Cálculo del área de un rectángulo

base = float(input("Dame la base: "))
altura = float(input("Dame la altura: "))

area = base * altura

print("La base es:", base)
print("La altura es:", altura)
print("El área del rectángulo es:", area)
```
